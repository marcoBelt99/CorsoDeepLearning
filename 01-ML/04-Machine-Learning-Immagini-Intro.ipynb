## Immagini e Pixel

Le immagini digitali sono costituite da una collezione ordinata di pixel che rappresentano dei punti individuali all'interno dell'immagine.

Ad ogni pixel è associato un colore o un valore di intensità attraverso un sistema di colore. Assegnando un colore ai singoli pixel di un'immagine si determina l'aspetto visivo dell'immagine.

Un sistema di colori è una rappresentazione astratta dello spazio dei colori. Ne esistono diversi per diversi spazi di colore e i componenti che rappresentano un sistema di colore sono chiamati "canali di colore".

- **Binary**: In un'immagine binaria, un pixel può essere solo nero o bianco, rappresentato con 0 o 1.
- **Grayscale**: In questa modalità, l'intensità del grigio è espressa con un unico valore numerico. Tipicamente, il range va da 0 (nero) a 255 (bianco).
- **RGB (Red, Green, Blu)**: Questo modello di colore usa tre canali, ognuno dei quali ha un range da 0 a 255 per rappresentare l'intensità del colore rosso, verde e blu del pixel.
- **RGBA (Red, Green, Blue, Alpha)**: Simile a RGB, ma con l'aggiunta di un canale alpha per rappresentare la trasparenza. I range per i canali RGB sono da 0 a 255, mentre per il canale alpha è tipicamente da 0 (completamente trasparente) a 255 (completamente opaco).
- **CMYB (Cyan, Magenta, Yellow, Black)**: Utilizzato nell'industria grafica, i canali CYMK hanno spesso range da 0 a 100% pre rappresentare le quantità di ciascun colore.
- **HSV e HSL (Hue, Saturation, Value/Lightness)**: La tonalità (H) la tinta tra i colori dell'arcobaleno (0=rosso,  255=violetto). La saturazione (S) indica quanto i colori sono "vividi" (0 = colori sbiaditi, 255 = colori accesi), e il valore (V) noto anche come luminosità (L) che va da 0 molto scuro a 255 molto chiaro.


Gli array numpy sono utili in quanto le immagini digitali sono una rappresentazione matriciale dei colori riportati su una matrice.

## Manipolazione dei Canali

Filtro Saturazione ?!

**Saturazione**: rappresenta l'intensità o la purezza dei colori in un'immagine.

```python
# il valore del fattore di saturazione tipicamente è compreso fra [-1, +inf]
# s = -1 completa desaturazione e si trasforma in una immagine nera.
# 5 = tra -1 e 0 indica una riduzione della saturazione.
# s = 1 corrisponde a l'attuale saturazione dell'immagine.
# s > 1 indica un aumento della saturazione.


saturation_factor = 1.5
saturated_imag = img_di_lavoro * saturation_factor

# Assicurati che i valori rimangano nel range 0, 255 e poi riportali in int
saturated_imag = np.clip(saturated_imag, 0, 255).astype(np.uint8)

plt.imshow(saturated_imag)
plt.axis('off') # nascondo gli assi
plt.title('Immagine saturazione')
plt.show()

plt.imshow(img_di_lavoro)
plt.axis('off') # nascondo gli assi
plt.title('Immagine originale')
plt.show()
```

Filtro Luminosità ?!

La luminosità è soggettiva e può variare da persona a persona. La luminanza no.

**Luminanza**: intensità luminosa di un'immagine o di un colore specifico. Viene misurata in (cd/m^2).

```python
# Aumento la luminosità aggiungendo un valore ai canali RGB

brightness_factor = 5
brightened_image = img_di_lavoro + brightness_factor # aumento di 5

# assicurati che i valori rimangano nel range 0-255
brightened_image = np.clip(brightened_image, 0, 255).astype(np.uint8) # clipping = assicurarmi che ogni elemento della mia matrici non superi il valore
# di 255 per evitare i limiti di rappresentazione

plt.imshow(brightened_image)
plt.axis('off') # nascondo gli assi
plt.title('Immagine luminosità')
plt.show()

print("\nE se volessi rendere l'immagine più scura?")
```

**Effetto polaroid**

Le Polaroid rendono i colori chiari più chiari e i colori scuri più scuri. Si può ottenere una buona approssimazione di questo effetto usando una variante della funzione logistica:

```python
def scaled_logistic(x):
  return 1 / (1 + np.exp( -12 * x + 6 ) )

x = np.linspace(0, 1, 100)
y = scaled_logistic(x)

plt.plot(x, y)
plt.plot(x, x, '--') # mostra le retta y = x per confronto
plt.title('Effetto Polaroid')
plt.show()
```

Infatti, la linea in arancione rappresenta l'andamento normale dei colori, mentre la linea blu è la funzione logistica, dove vado a modificare i valori dell'imamgine in base a tal funzione.

```python
def polaroid( image ):
  image = image.astype( np.float32 ) / 255 # Converto in range [0, 1]
  image = scaled_logistic( image ) #  applico la funzione all'immagine
  image = ( image * 255 ).astype( np.uint8 ) # convrto in range [0, 255]
  return image

polaroid_image = polaroid(img_di_lavoro)

plt.imshow(polaroid_image)
plt.axis('off') # nascondo gli assi
plt.title('Polaroid')
plt.show()
```

**Resize**

- Effettuo un resize per dare alle due immagini la stessa dimensione.
Una delle funzioni più utili che si usa anche nei modelli è la funzione di resize, perchè solitamente andare a lavorare con la dimensione originale nei modelli è molto costoso, quindi è importante sapere come andare a definire la dimenisone di larghezza e altezza.

```python
import cv2

# Attenzione: Open CV vuole prima la width e poi la height
new_shape = (img_di_lavoro.shape[1], img_di_lavoro.shape[0]) # mi salvo la shape dell'immagine di Oppenhaimmer
img_cane_lavoro = cv2.resize(img_cane_lavoro, new_shape) # porto l'immagine del cane alle stesse dimensioni di quella di Oppenhaimer

print(f"Dimensione della nuova immagine: {img_cane_lavoro.shape}")
print(f"Dimensione dell'immagine di Oppenhaimer: {img_di_lavoro.shape}\n")

plt.imshow( img_cane_lavoro )
plt.axis('off') # nascondo gli assi
plt.title('Cane resized')
plt.show()
```

Come si vede ho quindi fatto un upscaling dell'immagine, probabilmente andando a zoomare non ci sarà qualche pixel che non sarà benissimo definito.

**Interpolazione fra due immagini**

Visto che ho due immagini con le stesse dimensioni di larghezza e altezza, è possibile interpolarle per fare in modo da farle vedere una sovrapposta all'altra.

```python
def interpolate(image1, image2, coefficient):
  if image1.shape != image2.shape:
    raise ValueError("Le dimensioni delle immagini devono essere uguali. (Cioè devono avere la stessa shape).")

  # coefficiente serve per evitare di fare il flipping dell'immagine
  if coefficient < 0 or coefficient > 1:
    raise ValueError("Coefficiente deve essere tra [0,1]")

  interpolated = image1 * (1 - coefficient) + image2 * coefficient # Es) se coefficient = 0.5 vuol dire che ho metà immagine del cane e metà immagine di Openheimer
  # interpolated ora è un'immagine floating point, devo riconvertirla in intero
  return interpolated.astype(np.uint8)

# o = prendo solo image1, 1 = prendo solo image2
interpolated_image = interpolate(img_di_lavoro, img_cane_lavoro, 0.5)

plt.imshow(interpolated_image)
plt.axis('off') # nascondo gli assi
plt.title('Immagini interpolate')

plt.show()
```

L'effetto è stata di andare a prendere pixel per pixel quello che rappresentava in un'immagine, l'ho messo nella matrice e ho messo la restante parte in un'altra immagine.
Grazie a questa cosa riesco a sovrapporre le due immagini.

Creare figure geometriche più complesse

```python
# Per creare un poligono occorre definire la lista di punti (x, y) dei vertici del poligono
punti_triangolo = [(50, 50), (300, 150), (200, 250)] # caso triangolo

# trasformo la lista in un vettore numpy con shape ( 3, 1, 2 ).
# dove: 3 è il numero di vertici, 1 è perchè il vertice viene rappresentato da un solo punto, 2 per rappresentare le due coordinate x e y
tr_points = np.array( punti_triangolo, np.int32).reshape( len( punti_triangolo ), 1, 2 )

punti_poligono = [(800, 100), (950, 350), (900, 450), (800, 500)]
pol_points = np.array( punti_poligono, np.int32).reshape( len( punti_poligono ), 1, 2 ) #la shape quindi qui è 5, 1, 2

# cv.polylines(immagine, lista di poligonoi, True se deve chiudere i poligoni, colore, spessore)
polygon_image = cv2.polylines( img_cane_lavoro.copy(), [tr_points, pol_points], True, color_black, spessore )

# Stampa
plt.imshow( polygon_image )
plt.axis('off')
plt.title("Immagine con 2 figure geometriche")
plt.show()
```

**Maschera**

Sono dei segmenti che vengono posti su un'immagine, e sono usate per rappresentare delle porzioni di un'immagine che sono utili per compiti specifici.

Una maschera è un array multidimensionale di valori binari o in scala di grigi utilizzato per specificare le regioni di interesse in un'immagine o in un volume di dati.

In apprendimento automatico, viene usata per vari scopi, come:
- Nelle loss functions
- Segmentazione semantica
- Inpainting
- Filtraggio mirato

Vedo ora come rimuovere il testo dalla mia immagine

```python
# Esempio: se volessi ottenere una maschera che mi da il testo unife dall'immagine

# Creo una maschera basata sul confronto tra l'immagine originale e il colore rosso
color_red = [255, 0, 0]
mask = (text_image == color_red).all(axis=2).astype(np.uint8) # confronto l'img originale e l'immagine con il testo e seleziono solo i pixel in cui ho solo il rosso

print( mask )
print(f"La shape di mask è: {mask.shape}\n")

plt.imshow(mask)
plt.axis('off')
plt.title('Maschera del testo')
plt.show()
```

In pratica ho tutti 0 deve non c'è la scritta UniFe, invece dove c'è mi restituisce 1. Grazie a questa maschera e alla funzione Inpainting, che va a prendere i pixel nell'area circostante e fa un'interpolazione per cancellare la mia maschera, vado a cancellare la scritta dall'immagine, e ottengo l'imamgine originale nel seguente modo:

**Inpainting**

```python
# Eseguo l'inpainting
# inpaintRadius specifica quanti pixel l'operazione di inpainting deve riempire attorno all'area specificata dalla maschera.
#L'algoritmo di Telea per calcolare l'inpainting utilizza un approccio basato su equazioni alle derivate parziali (PDE)

image = text_image.copy().astype(np.uint8)
mask = mask.copy().astype(np.uint8)

inpainting_result = cv2.inpaint(image, mask, inpaintRadius=3, flags=cv2.INPAINT_TELEA)

# Stampa
image_list = [text_image, inpainting_result]
titles = ['Prima', 'Dopo']
fig, axes = plt.subplots(1, len(image_list), figsize=(16, 9)) # creo una figura con le dimensioni appropriate

for i, (image, title) in enumerate( zip(image_list, titles) ):
  axes[i].imshow(image)
  axes[i].axis('off')
  axes[i].set_title(title)

plt.subplots_adjust(wspace=0.005, hspace=0.2) # aggiusta lo spazio tra le immagini
plt.show()
```

L'immagine di destra non è l'immagine originale, ma è quella modificata in cui ho tolto la scritta UniFe.

## GRAYSCALE

La scala di grigi tratta principalmente immagini in bianco e nero ed è quindi formata da un solo canale per rappresentare l'immagine in bianco e nero. Ci sono varie formule per passare da RGB alla gray. Vedo la più comune.

```python
# Estraggo i canali dall'immagine
img = img_di_lavoro.copy()

# isolo i 3 canali
r = img[:,:,0]
g = img[:,:,1]
b = img[:,:,2]


# Nota: è equivalente a scrivere gray = np.mean(img, axis=2) axis=2 è l'asse dei colori
new_gray = (r//3 + g//3 + b//3) # faccio la media dei 3 canali

print(f"Shape della grayscale: {new_gray.shape}\n") # ha solo 1 canale adesso

plt.imshow(new_gray, cmap='gray') # matplotlib per stampare in vari formati necessita della color map, cioè della rappresentazione di colori con cui sto lavorando
plt.axis('off')
plt.title('Grayscale con la media')
plt.show()
```

```python
# Torno indietro

# Torno ai canali R, G, e B
r = new_gray
g = new_gray
b = new_gray

rgb_image = np.stack([r,g,b], axis=-1)

print( rgb_image.shape )

plt.imshow( rgb_image )
plt.axis('off')
plt.title('Immagine RGB')
plt.show()

# perchè non è a colori?
```

Perchè non è a colori?
Il motivo è che ho fatto la media, quindi ogni matrice avrà lo stesso valore, quindi andando a riaggiungerla sui vari canali andrò ad ottenere esattamente la copia dell'immagine in bianco e nero.
Se avessi voluto risalire all'immagine a colori avrei dovuto per forza salvare da qualche parte i vari canali.
